```python
X = D.drop(columns=["species"])
y = D["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Variables d'entrée :", X.columns.tolist())
print("Variable cible :", y.name)

print("\nRépartition des classes dans les données originales :")
print(y.value_counts(normalize=True))

print("\nRépartition dans l'ensemble d'entraînement :")
print(y_train.value_counts(normalize=True))

print("\nRépartition dans l'ensemble de test :")
print(y_test.value_counts(normalize=True))
```

```python
print("Dimensions de l'ensemble d'entraînement :")
print("X_train :", X_train.shape)
print("y_train :", y_train.shape)

print("\nDimensions de l'ensemble de test :")
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

print("\nProportion de l'entraînement :",
      len(X_train) / len(D) * 100, "%")

print("Proportion du test :",
      len(X_test) / len(D) * 100, "%")
```
