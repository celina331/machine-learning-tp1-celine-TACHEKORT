```python
import os
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from collections import Counter
```

```python
file_path = 'txt_file.txt'

# Création d'un fichier d'exemple si celui-ci n'existe pas encore
if not os.path.exists(file_path):
    with open(file_path, 'w', encoding='utf-8') as f:
        f.write("Introduction au Machine Learning avec Python.\nExploration du texte.\n")

# Ouverture et lecture avec l'encodage UTF-8
with open(file_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()
```

```python

print("=== CONTENU DU FICHIER TXT ===")
for line_num, line in enumerate(lines, start=1):
    clean_line = line.rstrip('\n')  
    print(f"Ligne {line_num}: {clean_line}")
```

```python

newsgroups_train = fetch_20newsgroups(subset='train', remove=('headers', 'footers', 'quotes'))
```

```python

num_documents = len(newsgroups_train.data)
num_labels = len(newsgroups_train.target)

print(f"Nombre de documents textuels : {num_documents}")
print(f"Nombre d'étiquettes (labels)  : {num_labels}")
```

```python

print("=== PREMIER DOCUMENT ===")
print(newsgroups_train.data[0][:300] + "...")  

target_idx = newsgroups_train.target[0]
print(f"\nIndex de la classe cible : {target_idx}")
```

```python

print("=== LISTE DES NOMS DE CLASSES ===")
for idx, name in enumerate(newsgroups_train.target_names):
    print(f"Code {idx:2d} -> {name}")

print(f"\nNom de la classe pour le 1er document : {newsgroups_train.target_names[target_idx]}")
```

```python

doc_lengths = [len(doc.split()) for doc in newsgroups_train.data]
avg_length = np.mean(doc_lengths)
print(f"Longueur moyenne des documents : {avg_length:.2f} mots")


words = newsgroups_train.data[0].lower().split()
most_common_words = Counter(words).most_common(5)
print("Top 5 des mots les plus fréquents dans le 1er document :", most_common_words)
```
