```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import (
    StandardScaler, MinMaxScaler, MaxAbsScaler, 
    RobustScaler, normalize, Binarizer,
    OrdinalEncoder, LabelEncoder, OneHotEncoder
)
```

```python
X_num = np.array([
    [1, -1,  2],
    [2,  0,  0],
    [0,  1, -1]
], dtype=float)
```

```python
#  StandardScaler (Standardisation Z-score)
std_scaler = StandardScaler()
X_std = std_scaler.fit_transform(X_num)

#  MinMaxScaler (Normalisation Min-Max [0, 1])
minmax_scaler = MinMaxScaler()
X_minmax = minmax_scaler.fit_transform(X_num)

#  MaxAbsScaler (Mise à l'échelle [-1, 1] par la valeur absolue max)
maxabs_scaler = MaxAbsScaler()
X_maxabs = maxabs_scaler.fit_transform(X_num)

#  RobustScaler (Basé sur la médiane et l'écart interquartile)
robust_scaler = RobustScaler()
X_robust = robust_scaler.fit_transform(X_num)

#  Normalisation L2 (par échantillon / ligne)
X_l2_norm = normalize(X_num, norm='l2')

print("=== 1. StandardScaler (Moyenne=0, Variance=1) ===")
print(np.round(X_std, 4))

print("\n=== 2. MinMaxScaler (Intervalle [0, 1]) ===")
print(np.round(X_minmax, 4))

print("\n=== 3. MaxAbsScaler (Intervalle [-1, 1]) ===")
print(np.round(X_maxabs, 4))

print("\n=== 4. RobustScaler (Médiane et IQR) ===")
print(np.round(X_robust, 4))

print("\n=== 5. Normalize L2 (Par échantillon / Ligne) ===")
print(np.round(X_l2_norm, 4))
```

```python
# Binarisation avec un seuil explicite (threshold = 0.0)
binarizer = Binarizer(threshold=0.0)
X_bin = binarizer.fit_transform(X_num)

print("=== Binarisation (Seuil threshold = 0.0) ===")
print(X_bin)
```

```python
df_cat = pd.DataFrame({
    'Feature 1': ['male', 'female', 'female', 'male'],
    'Feature 2': ['from US', 'from Europe', 'from Asia', 'from Africa'],
    'Feature 3': ['uses Safari', 'uses Safari', 'uses Firefox', 'uses Chrome'],
    'Target':    ['US', 'France', 'Japan', 'Algeria']
})

X_cat = df_cat[['Feature 1', 'Feature 2', 'Feature 3']]
y_cat = df_cat['Target']
```

```python
#  OrdinalEncoder sur les caractéristiques
ordinal_enc = OrdinalEncoder()
X_ordinal = ordinal_enc.fit_transform(X_cat)

print("=== 1. OrdinalEncoder pour les Features ===")
print("Matrice transformée :\n", X_ordinal)
print("\nAttribut categories_ :")
for col, cats in zip(X_cat.columns, ordinal_enc.categories_):
    print(f"  {col} -> {cats}")
```

```python
# LabelEncoder sur la variable cible
label_enc = LabelEncoder()
y_encoded = label_enc.fit_transform(y_cat)

print("=== 2. LabelEncoder pour la Target ===")
print("Target encodée :", y_encoded)
print("Mapping des classes :")
for code, label in enumerate(label_enc.classes_):
    print(f"  Code {code} -> {label}")
```
